## ⚙️ Paso 1: Configuración e Importación de Módulos

Cargamos la ruta de `src` para poder importar las clases personalizadas (`Jugador`, `Vehiculo`, `Mision`).

```python
import sys
import os
import json
from pathlib import Path
import matplotlib.pyplot as plt

# Configurar ruta al código fuente (src)
base_path = Path.cwd().resolve()
if base_path.name == 'notebooks':
    src_path = str(base_path.parent / 'src')
else:
    src_path = str(base_path / 'gta' / 'src')

if src_path not in sys.path:
    sys.path.append(src_path)

from motor import Jugador, Vehiculo, Mision
print('✅ Módulos del Proyecto GTA importados correctamente!')
```

## 👤 Paso 2: Creación del Jugador

Creamos a nuestro personaje con sus estadísticas clásicas de GTA: **vida**, **blindaje**, **dinero**, **estrellas de búsqueda policial** e **inventario**.

```python
# Instanciamos al protagonista
carlos = Jugador(nombre='Carlos M.', x=100.0, y=150.0, dinero=1500)

print(f'👤 Jugador: {carlos.nombre}')
print(f'❤️ Vida: {carlos.vida} | 🛡️ Blindaje: {carlos.blindaje}')
print(f'💵 Dinero en efectivo: ${carlos.dinero}')
print(f'⭐ Nivel de búsqueda policial: {carlos.estrellas_busqueda}')
print(f'🎒 Inventario: {carlos.inventario}')
print(f'📍 Posición en el mapa: (X={carlos.x}, Y={carlos.y})')
```

## 🏎️ Paso 3: Vehículos y Conducción

Instanciamos un coche deportivo clásico (**Banshee**) en una posición cercana. El jugador se sube, acelera, frena y se baja.

```python
# Crear un vehículo deportivo cerca del jugador
banshee = Vehiculo(nombre='Banshee V8', modelo='Banshee', color='Azul Eléctrico', x=105.0, y=152.0)

print(f'🚗 Coche aparcado: {banshee.modelo} ({banshee.color}) en (X={banshee.x}, Y={banshee.y})')

# El jugador se sube al coche
msg_subir = carlos.entrar_vehiculo(banshee)
print(f'\n👉 {msg_subir}')

# Simulamos aceleración y desplazamiento
banshee.acelerar(incremento=45.0)
banshee.mover(dx=30.0, dy=40.0)
print(f'⚡ Acelerando... Velocidad actual: {banshee.velocidad} km/h')
print(f'📍 Nueva posición del coche: (X={banshee.x}, Y={banshee.y})')

# Frenar y descender del vehículo
banshee.frenar(decremento=45.0)
msg_bajar = carlos.salir_vehiculo()
print(f'🛑 Coche detenido (Velocidad: {banshee.velocidad} km/h)')
print(f'👉 {msg_bajar}')
```

## 🎯 Paso 4: Sistema de Misiones y Recompensas

Definimos un sistema de encargos donde recibir una misión, cumplir el objetivo y cobrar la recompensa en efectivo.

```python
# Definir una misión inicial
mision_1 = Mision(
    id_mision='MIS-001',
    titulo='El Envío de Rafelbunyol',
    descripcion='Transporta un paquete de electrónica industrial sin llamar la atención de la policía.',
    recompensa_dinero=5000
)

print(f'📋 NUEVA MISIÓN: [{mision_1.titulo}]')
print(f'📝 Descripción: {mision_1.descripcion}')
print(f'💰 Recompensa: ${mision_1.recompensa_dinero}')
print(f'💵 Dinero actual antes de completar: ${carlos.dinero}')

# Completamos la misión
resultado = mision_1.completar(carlos)
print(f'\n🏆 {resultado}')
print(f'💵 Dinero actualizado de {carlos.nombre}: ${carlos.dinero}')
```

## 🗺️ Paso 5: Mini-Radar Gráfico 2D (Visualización con Matplotlib)

Generamos un mapa visual con cuadrícula urbana estilo **radar de GTA**, mostrando la posición del jugador, el coche y el punto de misión.

```python
fig, ax = plt.subplots(figsize=(8, 8), facecolor='#111827')
ax.set_facecolor('#1f2937')

# Calles principales (cuadrícula)
for calle_x in range(0, 301, 50):
    ax.axvline(calle_x, color='#374151', linestyle='--', linewidth=1.2, alpha=0.7)
for calle_y in range(0, 301, 50):
    ax.axhline(calle_y, color='#374151', linestyle='--', linewidth=1.2, alpha=0.7)

# Coordenadas clave
pos_jugador = (carlos.x, carlos.y)
pos_banshee = (banshee.x, banshee.y)
pos_mision = (250, 220)

# Marcadores en el radar
ax.scatter([pos_jugador[0]], [pos_jugador[1]], color='#3b82f6', s=180, edgecolors='white', linewidth=2, label='Jugador (Carlos)', zorder=5)
ax.scatter([pos_banshee[0]], [pos_banshee[1]], color='#ef4444', s=140, marker='s', edgecolors='yellow', linewidth=1.5, label='Vehículo (Banshee)', zorder=4)
ax.scatter([pos_mision[0]], [pos_mision[1]], color='#eab308', s=220, marker='*', edgecolors='white', linewidth=1.5, label='Punto de Entrega (Misión)', zorder=6)

# Estilo y etiquetas
ax.set_title('📍 MINI-RADAR GTA — MAPA URBANO Y PUNTOS DE INTERÉS', color='#f9fafb', fontsize=13, fontweight='bold', pad=15)
ax.set_xlim(-20, 320)
ax.set_ylim(-20, 320)
ax.set_xlabel('Coordenada X (Este-Oeste)', color='#9ca3af')
ax.set_ylabel('Coordenada Y (Norte-Sur)', color='#9ca3af')
ax.tick_params(colors='#9ca3af')
ax.legend(facecolor='#111827', edgecolor='#4b5563', labelcolor='#f3f4f6', loc='upper left')

plt.tight_layout()
plt.show()
```

## 💾 Paso 6: Guardar y Cargar Partida

Exportamos el estado actual en formato JSON dentro de `gta/data/estado_partida.json`.

```python
partida_data = {
    'jugador': {
        'nombre': carlos.nombre,
        'vida': carlos.vida,
        'blindaje': carlos.blindaje,
        'dinero': carlos.dinero,
        'estrellas': carlos.estrellas_busqueda,
        'posicion': {'x': carlos.x, 'y': carlos.y},
        'inventario': carlos.inventario
    },
    'vehiculo': {
        'modelo': banshee.modelo,
        'color': banshee.color,
        'posicion': {'x': banshee.x, 'y': banshee.y},
        'velocidad': banshee.velocidad
    },
    'misiones': [
        {
            'id': mision_1.id_mision,
            'titulo': mision_1.titulo,
            'completada': mision_1.completada
        }
    ]
}

data_dir = base_path if base_path.name != 'notebooks' else base_path.parent
archivo_guardado = data_dir / 'data' / 'estado_partida.json'
archivo_guardado.parent.mkdir(parents=True, exist_ok=True)

with open(archivo_guardado, 'w', encoding='utf-8') as f:
    json.dump(partida_data, f, indent=4, ensure_ascii=False)

print(f'💾 Partida guardada con éxito en: {archivo_guardado.name}')
print(json.dumps(partida_data, indent=2, ensure_ascii=False))
```
