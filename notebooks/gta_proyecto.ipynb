## 📡 Paso 1: Radar de Tendencias Virales de GTA en YouTube

Buscamos los Shorts más populares de GTA en YouTube para ver qué está funcionando ahora mismo.

```python
import sys
from pathlib import Path

# Configurar ruta a los módulos en src
base_dir = Path.cwd().resolve()
src_dir = base_dir.parent / "src" if base_dir.name == "notebooks" else base_dir / "gta" / "src"
if str(src_dir) not in sys.path:
    sys.path.append(str(src_dir))

from radar import buscar_tendencias_gta

# Término de búsqueda para encontrar los mejores Shorts
termino = "gta 5 stunts shorts viral"
print(f"🔍 Escaneando YouTube para: '{termino}'...")

shorts_encontrados = buscar_tendencias_gta(termino, limite=5)

print("\n🔥 TOP 5 SHORTS VIRALES ENCONTRADOS:")
print("=" * 60)
for i, short in enumerate(shorts_encontrados, 1):
    print(f"{i}. 📌 {short['titulo']}")
    print(f"   🔗 Enlace: {short['url']}")
    print(f"   ⏱️ Duración: {short['duracion']}s")
    print("-" * 60)
```

## 📥 Paso 2: Selección y Descarga del Clip

Elige el número del vídeo que más te guste de la lista anterior (o pega directamente cualquier enlace de YouTube Shorts que quieras transformar).

```python
from radar import descargar_video_sin_audio

# Elige el índice del vídeo (0 para el primero de la lista) o pega tu propia URL
indice_elegido = 0
url_a_descargar = shorts_encontrados[indice_elegido]['url'] if shorts_encontrados else "https://www.youtube.com/watch?v=ejemplo"

print(f"🎯 Vídeo seleccionado para transformar: {url_a_descargar}")

# Ruta donde guardaremos el clip de vídeo en bruto
carpeta_clips = base_dir.parent / "data" / "clips" if base_dir.name == "notebooks" else base_dir / "gta" / "data" / "clips"
archivo_video = carpeta_clips / "clip_viral_base.mp4"

print("⏳ Descargando el vídeo en alta calidad...")
ruta_final_video = descargar_video_sin_audio(url_a_descargar, str(archivo_video))
print(f"✅ Vídeo descargado exitosamente en: {ruta_final_video}")
```

## ✍️ Paso 3: Generador de Guiones con Gancho (Español & Inglés)

Creamos el guion narrativo que salvará el vídeo del baneo por contenido reutilizado y atrapará a la audiencia en los primeros 3 segundos.

```python
from guiones import obtener_guion

# Estilos disponibles: 'reto_rampa', 'tren_vs_todo', 'stunt_epico'
tipo_estilo = "reto_rampa"

guiones = obtener_guion(tipo=tipo_estilo)

print("🗣️ GUION EN ESPAÑOL:")
print(f"\"{guiones['es']}\"")
print("\n" + "=" * 50 + "\n")
print("🗣️ GUION EN INGLÉS (ENGLISH SCRIPT):")
print(f"\"{guiones['en']}\"")
```

## 🎙️ Paso 4: Generación de Voces Neuronales Gratuitas (`edge-tts`)

Generamos los archivos de voz en MP3 usando las voces neuronales de Microsoft en alta fidelidad: Álvaro (Español) y Christopher (Inglés).

```python
from IPython.display import Audio, display
from voces import sintetizar_voz

carpeta_audios = base_dir.parent / "data" / "audios" if base_dir.name == "notebooks" else base_dir / "gta" / "data" / "audios"

# 1. Generar audio en Español
ruta_audio_es = carpeta_audios / "locucion_es.mp3"
sintetizar_voz(guiones['es'], idioma="es", ruta_salida=str(ruta_audio_es))
print(f"✅ Audio en Español generado: {ruta_audio_es.name}")
display(Audio(str(ruta_audio_es)))

# 2. Generar audio en Inglés
ruta_audio_en = carpeta_audios / "locucion_en.mp3"
sintetizar_voz(guiones['en'], idioma="en", ruta_salida=str(ruta_audio_en))
print(f"✅ Audio en Inglés generado: {ruta_audio_en.name}")
display(Audio(str(ruta_audio_en)))
```
